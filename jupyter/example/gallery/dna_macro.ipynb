# Nucleosome DNA (1KX5)

Macro shot of the nucleosome DNA as backbone tubes and base sticks.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/dna_macro.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck

import urllib.request

with urllib.request.urlopen("https://files.rcsb.org/download/1KX5.pdb") as r:
    pdb = r.read().decode()
# Keep only the DNA strands (chains I and J)
data = "\n".join(line for line in pdb.splitlines()
                  if line[:6] in ("ATOM  ", "HETATM") and line[21] in "IJ") + "\nEND\n"
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.2,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'cartoon': True,
 'cartoonColor': 'chain',
 'cartoonAtoms': 'sidechains',
 'cartoonTubeRadius': 0.7,
 'ao': 1.0,
 'specular': 0.5,
 'gloss': 0.7,
 'shadows': 0.5,
 'rim': 0.15,
 'saturation': 1.15,
 'tonemap': True,
 'cartoonShade': 0.05,
 'cartoonQuality': 12,
 'atomShade': 0.2,
 'dofStrength': 1.3,
 'dofPosition': 0.15112}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.939693, 0.34202, 0, 0, -0.116978, 0.321394, 0.939693, 0, 0.321394, -0.883022, 0.34202, 0, 0, 0, 0, 1], "translation": [-12.999025495344796, -9.377188347280676], "zoom": 0.016164674000522926}

w = Speck(data=data, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "675px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("dna_macro.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#faf6f0")
```
