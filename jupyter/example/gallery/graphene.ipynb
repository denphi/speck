# Graphene

A graphene sheet at a grazing angle.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/graphene.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`, `ase`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck

from ase.build import graphene
atoms = graphene(size=(28, 28, 1), vacuum=0)
atoms.pbc = False  # needs: pip install ase
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.2,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'bondScale': 0.6,
 'ao': 1.0,
 'specular': 0.7,
 'gloss': 0.8,
 'shadows': 0.5,
 'rim': 0.45,
 'tonemap': True,
 'atomShade': 0.1,
 'bondShade': 0.15,
 'dofStrength': 1.3,
 'dofPosition': 0.401068,
 'unitCell': False,
 'atomColors': {'C': [0.23, 0.26, 0.31]}}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.866025, 0.5, 0, 0, -0.17101, 0.296198, -0.939693, 0, -0.469846, 0.813798, 0.34202, 0, 0, 0, 0, 1], "translation": [4.630941582557094, 0], "zoom": 0.021037998609445066}

w = Speck.from_ase(atoms, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "675px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("graphene.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#fcfcfd")
```
