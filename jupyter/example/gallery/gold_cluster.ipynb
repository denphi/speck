# Gold–thiolate cluster

A gold–thiolate nanocluster: reflective gold core, matte sulfur and hydrogen.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/gold_cluster.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.1`. The gallery version was composited on a gradient background.

```python
from ipyspeck import Speck

import urllib.request

with urllib.request.urlopen("https://raw.githubusercontent.com/denphi/speck/master/static/samples/au_thiol.xyz") as r:
    data = r.read().decode()
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'metallic': 1.0,
 'metallicAtoms': 'metals',
 'gloss': 0.8,
 'specular': 0.7,
 'atomShade': 0.08,
 'ao': 0.9,
 'shadows': 0.6,
 'rim': 0.3,
 'tonemap': True,
 'saturation': 1.1,
 'brightness': 0.55,
 'bondShade': 0.3,
 'dofPosition': 0.5,
 'unitCell': False}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.866025, 0, -0.5, 0, 0.17101, 0.939693, 0.296198, 0, 0.469846, -0.34202, 0.813798, 0, 0, 0, 0, 1], "translation": [0.1734697132873535, -0.427340030670166], "zoom": 0.04419057374628034}

w = Speck(data=data, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("gold_cluster.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#181e2e")
```
