# Heme in hemoglobin (4HHB)

Macro shot of the heme in hemoglobin's pocket, framed by blurred helices.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/heme_macro.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'cartoon': True,
 'cartoonColor': 'chain',
 'highlight': {'resName': 'HEM', 'chain': 'A'},
 'highlightScale': 1.35,
 'ao': 1.0,
 'specular': 0.5,
 'gloss': 0.7,
 'shadows': 0.5,
 'rim': 0.15,
 'saturation': 1.15,
 'tonemap': True,
 'cartoonShade': 0.05,
 'cartoonQuality': 14,
 'atomShade': 0.2,
 'dofStrength': 1.0,
 'dofPosition': 0.388873}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [-0.468695, -0.865275, 0.177831, 0, -0.85997, 0.492963, 0.132059, 0, -0.201931, -0.091034, -0.97516, 0, 0, 0, 0, 1], "translation": [-2.356826704983054, -3.448016240914814], "zoom": 0.04212519411058531}

w = Speck.from_pdb_id("4HHB", camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "675px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("heme_macro.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#fcfcfd")
```
