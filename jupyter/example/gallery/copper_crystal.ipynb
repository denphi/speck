# Copper crystal and unit cell

A copper crystal with its unit cell, metallic material and shadows.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/copper_crystal.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`, `ase`. The gallery version was composited on a gradient background.

```python
from ipyspeck import Speck

from ase.build import bulk
atoms = bulk('Cu', 'fcc', a=3.6, cubic=True).repeat((4, 4, 4))  # needs: pip install ase
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': False,
 'atomScale': 0.36,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'unitCell': True,
 'cellColor': '#8fa3c0',
 'cellRadius': 0.14,
 'metallic': 1.0,
 'gloss': 0.8,
 'specular': 0.7,
 'atomShade': 0.08,
 'ao': 1.0,
 'shadows': 0.6,
 'tonemap': True,
 'saturation': 1.1,
 'brightness': 0.55,
 'dofPosition': 0.5}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.819152, 0, -0.573576, 0, -0.242404, 0.906308, -0.346189, 0, 0.519837, 0.422618, 0.742404, 0, 0, 0, 0, 1], "translation": [0.9869265556335449, 0.9038739546203605], "zoom": 0.03748473093506708}

w = Speck.from_ase(atoms, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("copper_crystal.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#181e2e")
```
