# Streptavidin–biotin (1STP)

Macro shot of biotin in the streptavidin pocket.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/streptavidin_biotin.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'ao': 1.0,
 'specular': 0.5,
 'gloss': 0.7,
 'shadows': 0.5,
 'rim': 0.2,
 'saturation': 1.15,
 'tonemap': True,
 'cartoonShade': 0.05,
 'cartoonQuality': 12,
 'atomShade': 0.2,
 'cartoon': True,
 'cartoonColor': 'rainbow',
 'highlight': {'resName': 'BTN'},
 'highlightScale': 1.4,
 'dofStrength': 1.0,
 'dofPosition': 0.471355}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [-0.332583, -0.793358, -0.509874, 0, -0.887157, 0.446594, -0.116216, 0, 0.319908, 0.413687, -0.852363, 0, 0, 0, 0, 1], "translation": [2.6905863374041052, -1.2123172015236952], "zoom": 0.0671110007083637}

w = Speck.from_pdb_id("1STP", camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "675px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("streptavidin_biotin.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#f6fbf8")
```
