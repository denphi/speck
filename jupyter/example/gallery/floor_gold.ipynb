# Gold–thiolate cluster on the studio floor

The gold–thiolate cluster in metal on the studio floor.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/floor_gold.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.3`. The gallery version was composited on a gradient background.

```python
from ipyspeck import Speck

import urllib.request

with urllib.request.urlopen("https://raw.githubusercontent.com/denphi/speck/master/static/samples/au_thiol.xyz") as r:
    data = r.read().decode()
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.3,
 'relativeAtomScale': 1.0,
 'fxaa': 1,
 'metallic': 1.0,
 'metallicAtoms': 'metals',
 'gloss': 0.8,
 'specular': 0.75,
 'atomShade': 0.06,
 'ao': 1.0,
 'shadows': 0.6,
 'rim': 0.25,
 'tonemap': True,
 'saturation': 1.1,
 'brightness': 0.55,
 'floor': 1.0,
 'floorReflection': 0.35,
 'dofPosition': 0.5,
 'unitCell': False}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.939693, 0, -0.34202, 0, 0.059391, 0.984808, 0.163176, 0, 0.336824, -0.173648, 0.925417, 0, 0, 0, 0, 1], "translation": [0.1558704376220703, -4.561055507659912], "zoom": 0.02980789265250907}

w = Speck(data=data, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("floor_gold.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#fcfcfd")
```
