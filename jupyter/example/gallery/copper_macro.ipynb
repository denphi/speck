# Copper surface

Macro shot of a copper surface at a grazing angle; the rows of atoms recede into blur.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/copper_macro.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.1`, `ase`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck

from ase.build import bulk
atoms = bulk('Cu', 'fcc', a=3.6, cubic=True).repeat((16, 16, 2))  # needs: pip install ase
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': False,
 'atomScale': 0.42,
 'relativeAtomScale': 1.0,
 'fxaa': 1,
 'metallic': 1.0,
 'metallicAtoms': 'metals',
 'gloss': 0.8,
 'specular': 0.75,
 'atomShade': 0.06,
 'ao': 1.0,
 'shadows': 0.6,
 'rim': 0.25,
 'tonemap': True,
 'saturation': 1.1,
 'brightness': 0.55,
 'dofStrength': 1.3,
 'dofPosition': 0.353483,
 'unitCell': False}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.882948, 0.469472, 0, 0, -0.190951, 0.359127, -0.913545, 0, -0.428884, 0.806613, 0.406737, 0, 0, 0, 0, 1], "translation": [3.8294650021362324, -0.3232135772705078], "zoom": 0.02449686084371922}

w = Speck.from_ase(atoms, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "675px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("copper_macro.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#181e2e")
```
