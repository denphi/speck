# Carbon nanotube

A (10,10) carbon nanotube receding into blur.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/nanotube.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.1`, `ase`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck

from ase.build import nanotube
atoms = nanotube(10, 10, length=40, bond=1.42)
atoms.pbc = False  # needs: pip install ase
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.22,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'bondScale': 0.55,
 'ao': 1.0,
 'specular': 0.7,
 'gloss': 0.8,
 'shadows': 0.5,
 'rim': 0.5,
 'tonemap': True,
 'atomShade': 0.1,
 'bondShade': 0.2,
 'dofStrength': 1.3,
 'dofPosition': 0.385423,
 'unitCell': False}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.939693, 0.34202, 0, 0, -0.116978, 0.321394, -0.939693, 0, -0.321394, 0.883022, 0.34202, 0, 0, 0, 0, 1], "translation": [0.18005998144297397, 0.06137275695800781], "zoom": 0.02524960163367267}

w = Speck.from_ase(atoms, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "675px"
w
```

Custom element colors (run after the viewer is displayed):

```python
w.setAtomsColor({'C': [0.604, 0.651, 0.722]})
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("nanotube.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#181e2e")
```
