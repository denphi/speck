# AlphaFold RPP7 repeat domain

Macro shot of the leucine-rich-repeat domain of the AlphaFold RPP7 model.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/alphafold_macro.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.1`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'cartoon': True,
 'cartoonColor': 'plddt',
 'ao': 1.0,
 'brightness': 0.6,
 'specular': 0.55,
 'gloss': 0.65,
 'shadows': 0.55,
 'rim': 0.45,
 'saturation': 1.15,
 'tonemap': True,
 'cartoonShade': 0.05,
 'cartoonQuality': 14,
 'dofStrength': 1.2,
 'dofPosition': 0.23619}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.850879, -0.480318, 0.212838, 0, -0.519086, -0.706196, 0.481495, 0, -0.080966, -0.520175, -0.850214, 0, 0, 0, 0, 1], "translation": [4.966919747472421, -2.845077910851483e-06], "zoom": 0.020133201309552373}

w = Speck.from_alphafold("Q8W3K0", camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "675px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("alphafold_macro.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#181e2e")
```
