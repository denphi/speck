# Hemoglobin, glass surface (4HHB)

Hemoglobin's four chains as cartoons inside a translucent molecular surface, with the four hemes highlighted.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/hemoglobin_glass.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`. The gallery version was composited on a gradient background.

```python
from ipyspeck import Speck
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'cartoon': True,
 'cartoonColor': 'chain',
 'surface': True,
 'surfaceOpacity': 0.28,
 'surfaceColor': '#eef2f8',
 'surfaceResolution': 0.5,
 'highlight': {'resName': 'HEM'},
 'highlightScale': 1.35,
 'ao': 0.9,
 'specular': 0.45,
 'gloss': 0.7,
 'shadows': 0.4,
 'rim': 0.15,
 'saturation': 1.1,
 'tonemap': True,
 'cartoonShade': 0.05,
 'cartoonQuality': 12,
 'atomShade': 0.2,
 'dofPosition': 0.5}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.906308, 0, -0.422618, 0, 0, 1, 0, 0, 0.422618, 0, 0.906308, 0, 0, 0, 0, 1], "translation": [-0.6567592620849609, 0.052763938903808594], "zoom": 0.013107775522958942}

w = Speck.from_pdb_id("4HHB", camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("hemoglobin_glass.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#fcfcfd")
```
