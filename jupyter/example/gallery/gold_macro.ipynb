# Gold nanoparticle, 923 atoms

Macro shot of a 923-atom gold icosahedron: the front atoms are sharp, the rest melts into bokeh.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/gold_macro.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.1`, `ase`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck

from ase.cluster import Icosahedron
atoms = Icosahedron('Au', noshells=7)  # needs: pip install ase
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': False,
 'atomScale': 0.42,
 'relativeAtomScale': 1.0,
 'fxaa': 1,
 'metallic': 1.0,
 'metallicAtoms': 'metals',
 'gloss': 0.8,
 'specular': 0.75,
 'atomShade': 0.06,
 'ao': 1.0,
 'shadows': 0.6,
 'rim': 0.25,
 'tonemap': True,
 'saturation': 1.1,
 'brightness': 0.55,
 'dofStrength': 1.6,
 'dofPosition': 0.240134,
 'unitCell': False}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.882948, 0, -0.469472, 0, 0.145075, 0.951057, 0.272846, 0, 0.446494, -0.309017, 0.839733, 0, 0, 0, 0, 1], "translation": [9.690647332481767, 4.104791382617771], "zoom": 0.05568235225222479}

w = Speck.from_ase(atoms, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "675px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("gold_macro.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#181e2e")
```
