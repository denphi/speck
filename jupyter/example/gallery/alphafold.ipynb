# AlphaFold RPP7, by pLDDT

The AlphaFold model of RPP7, an *Arabidopsis* immune receptor, colored by prediction confidence (pLDDT).

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/alphafold.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`. The gallery version was composited on a gradient background.

```python
from ipyspeck import Speck
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'cartoon': True,
 'cartoonColor': 'plddt',
 'ao': 1.0,
 'brightness': 0.6,
 'specular': 0.55,
 'gloss': 0.65,
 'shadows': 0.55,
 'rim': 0.45,
 'fog': 0.45,
 'fogColor': '#181e2e',
 'saturation': 1.15,
 'tonemap': True,
 'cartoonShade': 0.05,
 'cartoonQuality': 12,
 'dofPosition': 0.5}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.939693, 0, 0.34202, 0, -0.059391, 0.984808, 0.163176, 0, -0.336824, -0.173648, 0.925417, 0, 0, 0, 0, 1], "translation": [0.4113143249511708, 1.6254193023681651], "zoom": 0.008646306532432639}

w = Speck.from_alphafold("Q8W3K0", camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("alphafold.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#181e2e")
```
