# Taxol (paclitaxel)

The anticancer drug Taxol (paclitaxel).

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/taxol.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.1`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck

import urllib.request

# Ideal coordinates of the TA1 ligand from the RCSB Chemical Component Dictionary
with urllib.request.urlopen("https://files.rcsb.org/ligands/view/TA1_ideal.sdf") as r:
    sdf = r.read().decode().splitlines()
n = int(sdf[3][:3])
rows = [line.split() for line in sdf[4:4 + n]]
data = "%d\nTA1\n" % n + "\n".join(
    "%s %s %s %s" % (r[3].capitalize(), r[0], r[1], r[2]) for r in rows)
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'ao': 1.0,
 'specular': 0.6,
 'gloss': 0.75,
 'shadows': 0.5,
 'rim': 0.2,
 'tonemap': True,
 'saturation': 1.15,
 'atomShade': 0.2,
 'dofStrength': 0.8,
 'dofPosition': 0.458754,
 'unitCell': False}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.939693, 0, -0.34202, 0, 0, 1, 0, 0, 0.34202, 0, 0.939693, 0, 0, 0, 0, 1], "translation": [-0.383939266204834, 0.25740718841552734], "zoom": 0.0625030652245151}

w = Speck(data=data, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("taxol.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#faf6f0")
```
