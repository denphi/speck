# Ubiquitin surface (1UBQ)

Ubiquitin's molecular surface colored N→C terminus, glossy with shadows.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/ubiquitin_surface.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`. The gallery version was composited on a gradient background.

```python
from ipyspeck import Speck
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'surface': True,
 'surfaceColor': 'rainbow',
 'surfaceResolution': 0.35,
 'ao': 1.0,
 'specular': 0.55,
 'gloss': 0.7,
 'shadows': 0.55,
 'rim': 0.2,
 'saturation': 1.15,
 'tonemap': True,
 'surfaceShade': 0.05,
 'dofPosition': 0.5,
 'cartoon': False,
 'cartoonColor': 'ss'}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.766044, 0, -0.642788, 0, -0.111619, 0.984808, -0.133022, 0, 0.633022, 0.173648, 0.754407, 0, 0, 0, 0, 1], "translation": [1.3143682479858398, 1.1387009620666504], "zoom": 0.02190360361491907}

w = Speck.from_pdb_id("1UBQ", camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("ubiquitin_surface.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#fcfcfd")
```
