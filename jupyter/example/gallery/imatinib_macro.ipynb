# Imatinib in ABL kinase (1IEP)

Macro shot of the leukemia drug imatinib (Gleevec) bound to ABL kinase.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/imatinib_macro.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'cartoon': True,
 'cartoonColor': 'rainbow',
 'highlight': {'resName': 'STI', 'chain': 'A'},
 'highlightScale': 1.4,
 'ao': 1.0,
 'specular': 0.5,
 'gloss': 0.7,
 'shadows': 0.45,
 'rim': 0.15,
 'saturation': 1.15,
 'tonemap': True,
 'cartoonShade': 0.05,
 'cartoonQuality': 14,
 'atomShade': 0.15,
 'dofStrength': 1.0,
 'dofPosition': 0.409365}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.987187, 0.127623, -0.095783, 0, 0.006122, -0.630112, -0.77648, 0, -0.15945, 0.765945, -0.62282, 0, 0, 0, 0, 1], "translation": [7.241221921425881, -6.408849859393055], "zoom": 0.04986258404208556}

w = Speck.from_pdb_id("1IEP", camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "675px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("imatinib_macro.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#faf6f0")
```
