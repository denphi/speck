# Caffeine

Caffeine in an outlined toon style.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/caffeine.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.1`. The gallery version was composited on a gradient background.

```python
from ipyspeck import Speck

import urllib.request

with urllib.request.urlopen("https://raw.githubusercontent.com/denphi/speck/master/static/samples/caffeine.xyz") as r:
    data = r.read().decode()
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'ao': 0.4,
 'outline': 1.0,
 'outlineWidth': 1.8,
 'outlineColor': '#1d2233',
 'atomShade': 0.1,
 'saturation': 1.25,
 'specular': 0.4,
 'gloss': 0.6,
 'dofPosition': 0.5,
 'unitCell': False}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.939693, 0, -0.34202, 0, 0.116978, 0.939693, 0.321394, 0, 0.321394, -0.34202, 0.883022, 0, 0, 0, 0, 1], "translation": [-0.24829018115997314, 0.31167101860046387], "zoom": 0.12727190517385611}

w = Speck(data=data, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("caffeine.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#fcfcfd")
```
