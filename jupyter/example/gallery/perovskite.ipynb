# SrTiO₃ perovskite

The SrTiO₃ perovskite in its unit cell, with metallic Sr and Ti.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/perovskite.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.1`, `ase`. The gallery version was composited on a gradient background.

```python
from ipyspeck import Speck

from ase.spacegroup import crystal
atoms = crystal(['Sr', 'Ti', 'O'], basis=[(0, 0, 0), (0.5, 0.5, 0.5), (0.5, 0.5, 0)],
                spacegroup=221, cellpar=[3.905, 3.905, 3.905, 90, 90, 90]).repeat((3, 3, 3))  # needs: pip install ase
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': False,
 'atomScale': 0.34,
 'relativeAtomScale': 1.0,
 'fxaa': 1,
 'metallic': 1.0,
 'metallicAtoms': 'metals',
 'gloss': 0.8,
 'specular': 0.75,
 'atomShade': 0.06,
 'ao': 1.0,
 'shadows': 0.6,
 'rim': 0.25,
 'tonemap': True,
 'saturation': 1.1,
 'brightness': 0.55,
 'unitCell': True,
 'cellColor': '#8fa3c0',
 'cellRadius': 0.1,
 'dofPosition': 0.5}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.819152, 0, -0.573576, 0, -0.242404, 0.906308, -0.346189, 0, 0.519837, 0.422618, 0.742404, 0, 0, 0, 0, 1], "translation": [0.7509766960144031, 0.4591413879394519], "zoom": 0.04514831682699209}

w = Speck.from_ase(atoms, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Custom element colors (run after the viewer is displayed):

```python
w.setAtomsColor({'O': [0.886, 0.251, 0.227], 'Sr': [0.435, 0.518, 0.847], 'Ti': [0.851, 0.867, 0.89]})
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("perovskite.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#181e2e")
```
