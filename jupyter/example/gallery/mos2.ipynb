# MoS₂ monolayer

A MoS₂ monolayer: metallic molybdenum, yellow sulfur.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/mos2.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`, `ase`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck

from ase.build import mx2
atoms = mx2('MoS2', size=(18, 18, 1))
atoms.pbc = False  # needs: pip install ase
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': False,
 'atomScale': 0.36,
 'relativeAtomScale': 1.0,
 'fxaa': 1,
 'metallic': 1.0,
 'metallicAtoms': 'metals',
 'gloss': 0.8,
 'specular': 0.75,
 'atomShade': 0.06,
 'ao': 1.0,
 'shadows': 0.6,
 'rim': 0.25,
 'tonemap': True,
 'saturation': 1.1,
 'brightness': 0.55,
 'dofStrength': 1.3,
 'dofPosition': 0.405814,
 'unitCell': False}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.906308, 0.422618, 0, 0, -0.158316, 0.339509, -0.927184, 0, -0.391845, 0.840314, 0.374607, 0, 0, 0, 0, 1], "translation": [5.145080093231201, 0.28571271133422904], "zoom": 0.018602993160779905}

w = Speck.from_ase(atoms, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "675px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("mos2.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#181e2e")
```
