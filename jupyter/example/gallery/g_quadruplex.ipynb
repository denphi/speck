# G-quadruplex with K⁺ (1KF1)

A DNA G-quadruplex seen down its axis, with its K⁺ ion.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/g_quadruplex.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`. The gallery version was composited on a gradient background.

```python
from ipyspeck import Speck
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.2,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'ao': 1.0,
 'specular': 0.5,
 'gloss': 0.7,
 'shadows': 0.5,
 'rim': 0.45,
 'saturation': 1.15,
 'tonemap': True,
 'cartoonShade': 0.05,
 'cartoonQuality': 12,
 'atomShade': 0.2,
 'cartoon': True,
 'cartoonColor': 'rainbow',
 'cartoonAtoms': 'sidechains',
 'cartoonTubeRadius': 0.6,
 'highlight': {'resName': 'K'},
 'highlightColor': '#b18cff',
 'highlightScale': 2.4,
 'dofPosition': 0.5}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.150151, -0.01247, 0.988584, 0, -0.01247, 0.999817, 0.014505, 0, -0.988584, -0.014505, 0.149968, 0, 0, 0, 0, 1], "translation": [-0.20835208892822266, 1.4964327812194842], "zoom": 0.023787121655648318}

w = Speck.from_pdb_id("1KF1", camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("g_quadruplex.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#181e2e")
```
