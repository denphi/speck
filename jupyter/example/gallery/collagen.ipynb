# Collagen triple helix (1BKV)

The collagen triple helix as a surface colored by strand, with a shallow depth of field.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/collagen.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'ao': 1.0,
 'specular': 0.5,
 'gloss': 0.7,
 'shadows': 0.5,
 'rim': 0.2,
 'saturation': 1.15,
 'tonemap': True,
 'cartoonShade': 0.05,
 'cartoonQuality': 12,
 'atomShade': 0.2,
 'surface': True,
 'surfaceColor': 'chain',
 'surfaceResolution': 0.35,
 'surfaceShade': 0.05,
 'ligands': False,
 'dofStrength': 1.2,
 'dofPosition': 0.412271,
 'cartoon': False,
 'cartoonColor': 'ss'}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0, 0, -1, 0, 0.866025, 0.5, 0, 0, 0.5, -0.866025, 0, 0, 0, 0, 0, 1], "translation": [6.1662213155627255, 2.206148624420166], "zoom": 0.01947714833824872}

w = Speck.from_pdb_id("1BKV", camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "675px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("collagen.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#faf6f0")
```
