# Chlorophyll a

Chlorophyll a, seen face-on, with its central magnesium highlighted.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/chlorophyll.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.2`. The gallery version was composited on a gradient background.

```python
from ipyspeck import Speck

import urllib.request

# Ideal coordinates of the CLA ligand from the RCSB Chemical Component Dictionary
with urllib.request.urlopen("https://files.rcsb.org/ligands/view/CLA_ideal.sdf") as r:
    sdf = r.read().decode().splitlines()
n = int(sdf[3][:3])
rows = [line.split() for line in sdf[4:4 + n]]
data = "%d\nCLA\n" % n + "\n".join(
    "%s %s %s %s" % (r[3].capitalize(), r[0], r[1], r[2]) for r in rows)
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'ao': 1.0,
 'specular': 0.6,
 'gloss': 0.75,
 'shadows': 0.5,
 'rim': 0.2,
 'tonemap': True,
 'saturation': 1.15,
 'atomShade': 0.2,
 'highlight': {'element': 'Mg'},
 'highlightColor': '#35c24f',
 'highlightScale': 1.6,
 'dofStrength': 0.6,
 'dofPosition': 0.391452,
 'unitCell': False}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.009027, -0.630385, 0.77623, 0, -0.071153, 0.773889, 0.629311, 0, -0.997425, -0.060912, -0.037868, 0, 0, 0, 0, 1], "translation": [-4.712923287984727e-07, -4.371616593780064], "zoom": 0.037718491501736395}

w = Speck(data=data, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("chlorophyll.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#f6fbf8")
```
