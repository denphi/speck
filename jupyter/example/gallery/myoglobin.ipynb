# Myoglobin, toon style (1MBN)

Myoglobin, the first protein structure ever solved, in an outlined toon style.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/myoglobin.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.1`. The gallery version was composited on a gradient background.

```python
from ipyspeck import Speck
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.24,
 'relativeAtomScale': 0.64,
 'fxaa': 1,
 'cartoon': True,
 'cartoonColor': 'rainbow',
 'ao': 0.4,
 'outline': 1.0,
 'outlineWidth': 1.6,
 'outlineColor': '#1d2233',
 'cartoonShade': 0.0,
 'atomShade': 0.1,
 'saturation': 1.2,
 'cartoonQuality': 12,
 'specular': 0.3,
 'gloss': 0.6,
 'dofPosition': 0.5}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.866025, 0, -0.5, 0, 0, 1, 0, 0, 0.5, 0, 0.866025, 0, 0, 0, 0, 1], "translation": [-2.540024757385254, -0.24352219757080107], "zoom": 0.022567542696829646}

w = Speck.from_pdb_id("1MBN", camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("myoglobin.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#fcfcfd")
```
