# C₆₀ buckminsterfullerene

C₆₀ buckminsterfullerene as glossy blue spheres, with a shallow depth of field.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/gallery/buckyball.jpg" width="360"/>

This notebook reproduces the gallery image with the same settings and camera. Requires `ipyspeck>=0.8.1`, `ase`. The gallery version was composited on a gradient background with a light vignette.

```python
from ipyspeck import Speck

from ase.build import molecule
atoms = molecule('C60')  # needs: pip install ase
```

```python
settings = {'aoRes': 1024,
 'aoSamples': 1024,
 'spf': 64,
 'bonds': True,
 'atomScale': 0.3,
 'relativeAtomScale': 1.0,
 'fxaa': 1,
 'bondScale': 0.45,
 'ao': 1.0,
 'specular': 0.8,
 'gloss': 0.85,
 'shadows': 0.5,
 'rim': 0.5,
 'tonemap': True,
 'atomShade': 0.05,
 'bondShade': 0.1,
 'dofStrength': 1.2,
 'dofPosition': 0.382667,
 'unitCell': False}

# Camera of the gallery image (rotation, pan, zoom)
camera = {"rotation": [0.866025, 0, -0.5, 0, 0.17101, 0.939693, 0.296198, 0, 0.469846, -0.34202, 0.813798, 0, 0, 0, 0, 1], "translation": [-0.00917688070868139, 0.5774328167693595], "zoom": 0.13506964809078278}

w = Speck.from_ase(atoms, camera=camera, **settings)
# Same aspect ratio as the image, so the framing matches
w.layout.width, w.layout.height = "540px", "540px"
w
```

Custom element colors (run after the viewer is displayed):

```python
w.setAtomsColor({'C': [0.184, 0.435, 0.878]})
```

Save a high-resolution PNG (rendered in the browser; the file appears a moment after the cell runs):

```python
w.save_image("buckyball.png", width=1080, supersample=2, aoRes=2048, transparent=False,
             background="#181e2e")
```
