# AlphaFold models with ipyspeck

Cartoon and molecular-surface rendering of an AlphaFold prediction, colored by the model's per-residue confidence (pLDDT).

Example: [Q8W3K0](https://alphafold.ebi.ac.uk/entry/Q8W3K0), RPP7 (Recognition of *Peronospora parasitica* 7), an *Arabidopsis thaliana* immune receptor of 1138 residues.

```python
import json
import urllib.request

from ipyspeck import Speck


def fetch_alphafold(uniprot_id):
    """Download the latest AlphaFold DB model for a UniProt accession as PDB text."""
    api = f"https://alphafold.ebi.ac.uk/api/prediction/{uniprot_id}"
    with urllib.request.urlopen(api) as r:
        entry = json.load(r)[0]
    with urllib.request.urlopen(entry["pdbUrl"]) as r:
        return r.read().decode()


pdb = fetch_alphafold("Q8W3K0")
```

## Cartoon colored by confidence

AlphaFold stores pLDDT in the B-factor column: dark blue > 90, light blue 70–90, yellow 50–70, orange < 50. The file has no HELIX/SHEET records, so secondary structure is computed from backbone hydrogen bonds.

```python
cartoon = Speck(data=pdb, cartoon=True, cartoonColor="plddt")
cartoon.layout.height = "600px"
cartoon
```

## Molecular surface

Solvent-excluded surface with a 1.4 Å probe. `surfaceResolution` is the grid spacing in Å (smaller is smoother and slower).

```python
surface = Speck(data=pdb, surface=True, surfaceColor="plddt")
surface.layout.height = "600px"
surface
```

## Styling for figures

Glossy surface colored by element, with a soft outline:

```python
surface.surfaceColor = "element"
surface.specular = 0.5
surface.gloss = 0.6
surface.outline = 0.3
```
