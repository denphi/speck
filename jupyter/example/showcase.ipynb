# ipyspeck showcase

Every setting is a constructor argument and can be changed live; the toolbar (top right) toggles cartoon, surface and ligands, switches styles and views, and saves a PNG. Structures are downloaded from RCSB and AlphaFold DB.

```python
from ipyspeck import Speck
import ipywidgets as widgets
```

## Cartoon with highlighted ligands

Hemoglobin, colored by chain, with the four hemes enlarged and the rest ghosted. `cover` combines shadows, rim light, fog and gloss.

```python
hb = Speck.from_pdb_id('4HHB', cartoonColor='chain', highlight={'resName': 'HEM'},
                        highlightScale=1.3, ghost=0.55)
hb.apply_preset('cover')
hb.layout.height = '480px'
hb
```

## Transparent surface

The molecular surface as glass over the cartoon; the surface keeps its own ambient occlusion.

```python
glass = Speck.from_pdb_id('4HHB', cartoonColor='chain')
glass.apply_preset('glass')
glass.layout.height = '480px'
glass
```

```python
# Change it live
glass.surfaceOpacity = 0.2
glass.surfaceColor = 'chain'
```

## AlphaFold model colored by confidence (pLDDT)

```python
af = Speck.from_alphafold('Q8W3K0', shadows=0.5, rim=0.3)
af.layout.height = '480px'
af
```

```python
af.surface = True
af.surfaceColor = 'plddt'
```

## Metals with a unit cell

A copper crystal from ASE (needs `pip install ase`): metallic reflections only on metal atoms, with the unit cell drawn.

```python
from ase.build import bulk
cu = Speck.from_ase(bulk('Cu', 'fcc', a=3.6, cubic=True).repeat((3, 3, 3)),
                    atomScale=0.35, bonds=False)
cu.apply_preset('metal')
cu.shadows = 0.5
cu.layout.height = '420px'
cu
```

## Trajectory

An NMR ensemble (10 models) with play controls linked to `frame`.

```python
nmr = Speck.from_pdb_id('1D3Z', cartoonColor='rainbow', specular=0.4)
nmr.layout.height = '420px'
widgets.VBox([nmr, nmr.trajectory_controls(interval=400)])
```

## Figures and animations

Rendered in the browser at high resolution after the cell runs; the files appear a moment later. Rotate the hemoglobin view above first: exports use the current camera.

```python
hb.save_image('hemoglobin.png', width=2400)
hb.save_animation('hemoglobin.gif', frames=36, fps=12)
```
