# Your first video in three steps

Make a short, high-quality video of a protein: show it, watch the video, save it.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/videos/tour.gif" width="360"/>

Needs `ipyspeck>=0.8.4` and a recent Chrome, Edge, Safari (16.4+) or Firefox (130+), which make the video. Run the cells from the top, one at a time.

**Step 1: show a structure.** Any PDB ID works; here, hemoglobin (4HHB) as a cartoon.

```python
from ipyspeck import Speck

w = Speck.from_pdb_id("4HHB", cartoon=True)
w
```

**Step 2: watch the video.** A player appears in the viewer: pause it, drag the time slider,
or close it with **×**. Nothing is saved yet.

`"tour"` turns the protein, flies in to its ligand (here the heme) and comes back.
Other ready-made videos: `"spin"`, `"rock"`, `"orbit"`, `"focus"`, `"reveal"` (cut open),
`"trajectory"` and `"showcase"`; see notebook 02.

```python
w.preview("tour")
```

**Step 3: save it.** The video is made in the viewer (watch its progress bar) and
`hemoglobin.mp4` appears next to this notebook when it is done, after a minute or so.

```python
w.save_video("hemoglobin.mp4", "tour", title="Hemoglobin", subtitle="PDB 4HHB")
```

### Without code: the video button

The clapperboard button at the right of the viewer's toolbar lists the same videos and sizes.
Pick one, watch it, then press **Save video** in the player: the MP4 goes to your browser's
downloads folder.

### Common changes

| You want | Add |
|---|---|
| A longer or shorter video | `seconds=20` |
| A video for slides (default), phones or social media | `size="1080p"`, `size="vertical"`, `size="square"` |
| A quick test first | `quality="draft"` |
| The best quality (slower) | `quality="best"`, `size="4k"` |
| A dark background | `background=["#1b2330", "#07090d"]` (top and bottom colors) |
| A GIF for a web page | a file name ending in `.gif` (needs `pip install "ipyspeck[video]"`) |

```python
# A vertical video for phones, on a dark background
w.save_video("hemoglobin_phone.mp4", "spin", size="vertical", background=["#1b2330", "#07090d"])
```

### If something does not work

- **Nothing happens after `save_video`**: the viewer must be on screen (scroll up to it) while the
  video is made; its progress bar shows there. The file appears when the bar finishes.
- **"cannot encode video"**: use a recent Chrome, Edge, Safari or Firefox, or save a GIF instead:
  `w.save_video("movie.gif", "spin")` (install what GIFs need with `pip install "ipyspeck[video]"`).
- **Too slow**: add `quality="draft"` for a quick test, or use `size="720p"`.
- **To stop a video being made**, press **Cancel** in the viewer.
