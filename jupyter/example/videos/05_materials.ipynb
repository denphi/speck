# Materials: clusters and crystals

A gold–thiolate nanocluster in a metal look: a slow spin, then a cut through its core.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/videos/reveal_gold.gif" width="360"/>

Needs `ipyspeck>=0.8.4` and a recent Chrome, Edge, Safari (16.4+) or Firefox (130+), which make the video. Run the cells from the top, one at a time.

**Step 1: load the cluster** (an XYZ file) and give it a metal look.

```python
import urllib.request
from ipyspeck import Speck

url = "https://raw.githubusercontent.com/denphi/speck/master/static/samples/au_thiol.xyz"
data = urllib.request.urlopen(url).read().decode()

w = Speck(data=data)
w.apply_preset("metal")
w
```

**Step 2: a spin on a dark background.**

```python
w.preview("spin", background=["#20242c", "#050608"])
```

```python
w.save_video("gold_spin.mp4", "spin", background=["#20242c", "#050608"], vignette=0.3)
```

**Step 3: cut it open** to show the gold core inside the ligand shell.

```python
w.save_video("gold_cut.mp4", "reveal", seconds=10, background=["#20242c", "#050608"], title="Au–thiolate cluster")
```

### Your own structures

`Speck.from_ase(atoms)`, `Speck.from_pymatgen(structure)` and `Speck.from_file("POSCAR.xyz")`
load materials the same way; periodic structures show their unit cell. Then use any
ready-made video.

### If something does not work

- **Nothing happens after `save_video`**: the viewer must be on screen (scroll up to it) while the
  video is made; its progress bar shows there. The file appears when the bar finishes.
- **"cannot encode video"**: use a recent Chrome, Edge, Safari or Firefox, or save a GIF instead:
  `w.save_video("movie.gif", "spin", size="480p")` (needs the `Pillow` package).
- **Too slow**: add `quality="draft"` for a quick test, or use `size="720p"`.
- **To stop a video being made**, press **Cancel** in the viewer.
