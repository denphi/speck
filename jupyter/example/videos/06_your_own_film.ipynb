# Your own film

When the ready-made videos are not enough, put shots together like a storyboard: each shot
starts where the previous one ended.

Needs `ipyspeck>=0.8.4` and a recent Chrome, Edge, Safari (16.4+) or Firefox (130+), which make the video. Run the cells from the top, one at a time.

```python
from ipyspeck import Speck, shots

w = Speck.from_pdb_id("4HHB", cartoon=True)
w
```

### A storyboard

| Shot | What it does |
|---|---|
| `shots.title("text", seconds)` | text over the picture (use with `together`) |
| `shots.turntable(seconds)`, `shots.rock(...)`, `shots.orbit(...)` | turn the structure |
| `shots.zoom(seconds, factor=2)` | move closer |
| `shots.fly_to(selection, seconds)` | center a part and fill the picture with it |
| `shots.rack_focus(selection, seconds)` | focus on a part, blurring the rest |
| `shots.cut_open(seconds)` | slice the structure open |
| `shots.fade(seconds, setting=value)` | change a setting smoothly, e.g. `surfaceOpacity=0.3` |
| `shots.crossfade(seconds, setting=value)` | dissolve into new settings, e.g. `surface=True` |
| `shots.hold(seconds)` | pause |
| `shots.home(seconds)` | back to the first view |
| `shots.together(a, b)` | two shots at the same time |

```python
heme = {"resName": "HEM", "chain": "A"}

film = [
    shots.together(shots.orbit(6, degrees=180), shots.title("Hemoglobin", subtitle="carries oxygen")),
    shots.crossfade(2, surface=True, surfaceOpacity=0.35),      # a glass surface appears
    shots.fly_to(heme, 3),                                     # into the heme pocket
    shots.together(shots.rock(4, degrees=15), shots.title("Heme", seconds=4, subtitle="binds O₂")),
    shots.home(3),
]
w.preview(film)
```

```python
w.save_video("hemoglobin_story.mp4", film)
```

### Moves you pose yourself (keyframes)

1. Run the next cell: it remembers the current view.
2. Rotate and zoom the viewer to the next view with the mouse, then run the cell after it.
3. Repeat, then play the path through your views.

```python
keys = [w.keyframe()]              # view 1
```

```python
keys.append(w.keyframe())          # view 2: run after moving the viewer
```

```python
keys.append(w.keyframe(fog=0.4))   # view 3, with fog reaching 0.4 there
```

```python
w.preview([shots.keyframes(keys, seconds=8)])
```

### If something does not work

- **Nothing happens after `save_video`**: the viewer must be on screen (scroll up to it) while the
  video is made; its progress bar shows there. The file appears when the bar finishes.
- **"cannot encode video"**: use a recent Chrome, Edge, Safari or Firefox, or save a GIF instead:
  `w.save_video("movie.gif", "spin")` (install what GIFs need with `pip install "ipyspeck[video]"`).
- **Too slow**: add `quality="draft"` for a quick test, or use `size="720p"`.
- **To stop a video being made**, press **Cancel** in the viewer.
