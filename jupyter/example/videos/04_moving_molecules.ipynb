# Moving molecules: ensembles and trajectories

Play the frames of a structure smoothly: an NMR ensemble from the PDB, or your own MD simulation.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/videos/trajectory.gif" width="360"/>

Needs `ipyspeck>=0.8.4` and a recent Chrome, Edge, Safari (16.4+) or Firefox (130+), which make the video. Run the cells from the top, one at a time.

**Step 1: an NMR ensemble.** Ubiquitin (1D3Z) has 10 models, which become frames.

```python
from ipyspeck import Speck

w = Speck.from_pdb_id("1D3Z", cartoon=True)
w
```

**Step 2: watch and save.** The frames are blended into one another, so the motion is smooth
even with few frames. `seconds=` sets how long the whole trajectory takes.

```python
w.preview("trajectory")
```

```python
w.save_video("ubiquitin_ensemble.mp4", "trajectory", seconds=10, title="Ubiquitin", subtitle="NMR ensemble, 1D3Z")
```

### If something does not work

- **Nothing happens after `save_video`**: the viewer must be on screen (scroll up to it) while the
  video is made; its progress bar shows there. The file appears when the bar finishes.
- **"cannot encode video"**: use a recent Chrome, Edge, Safari or Firefox, or save a GIF instead:
  `w.save_video("movie.gif", "spin", size="480p")` (needs the `Pillow` package).
- **Too slow**: add `quality="draft"` for a quick test, or use `size="720p"`.
- **To stop a video being made**, press **Cancel** in the viewer.
