# Ready-made videos

Each video is one line. They adapt to the structure: `tour` and `focus` find its ligand,
`trajectory` plays its frames. Run the first cell, then any of the others.

Needs `ipyspeck>=0.8.4` and a recent Chrome, Edge, Safari (16.4+) or Firefox (130+), which make the video. Run the cells from the top, one at a time.

```python
from ipyspeck import Speck

w = Speck.from_pdb_id("4HHB", cartoon=True)
w
```

## Spin: `"spin"`

One full turn. It loops seamlessly, so it can play over and over on a slide. 

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/videos/spin.gif" width="300"/>

```python
w.preview("spin")
# w.save_video("spin.mp4", "spin")
```

## Rock: `"rock"`

A gentle swing back and forth; also loops seamlessly. 

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/videos/rock.gif" width="300"/>

```python
w.preview("rock")
# w.save_video("rock.mp4", "rock")
```

## Orbit: `"orbit"`

A full turn around a tilted axis, so the top and bottom come into view. 

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/videos/orbit.gif" width="300"/>

```python
w.preview("orbit")
# w.save_video("orbit.mp4", "orbit")
```

## Ligand tour: `"tour"`

Turns, flies in to the ligand, looks around it and returns to the start. It visits the largest ligand; `target=` picks another part (see notebook 03).

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/videos/tour.gif" width="300"/>

```python
w.preview("tour")
# w.save_video("tour.mp4", "tour")
```

## Focus on the ligand: `"focus"`

Moves close to the ligand and pulls the focus onto it, like a macro photograph. 

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/videos/focus.gif" width="300"/>

```python
w.preview("focus")
# w.save_video("focus.mp4", "focus")
```

## Cut open: `"reveal"`

Slices the structure open while it turns, to show what is inside. 

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/videos/reveal.gif" width="300"/>

```python
w.preview("reveal")
# w.save_video("reveal.mp4", "reveal")
```

## Showcase: `"showcase"`

Half a turn, then the ligand up close with a focus pull (or the inside, when there is no ligand), and back.

## Play trajectory: `"trajectory"`

For structures with several frames (NMR ensembles, MD trajectories); see notebook 04.

## Length and titles

Every video takes `seconds=` and an optional `title=` / `subtitle=` shown at the start:

```python
w.save_video("hemoglobin_orbit.mp4", "orbit", seconds=15, title="Hemoglobin", subtitle="Oxygen transport")
```

### If something does not work

- **Nothing happens after `save_video`**: the viewer must be on screen (scroll up to it) while the
  video is made; its progress bar shows there. The file appears when the bar finishes.
- **"cannot encode video"**: use a recent Chrome, Edge, Safari or Firefox, or save a GIF instead:
  `w.save_video("movie.gif", "spin", size="480p")` (needs the `Pillow` package).
- **Too slow**: add `quality="draft"` for a quick test, or use `size="720p"`.
- **To stop a video being made**, press **Cancel** in the viewer.
