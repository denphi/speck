# A drug in its binding pocket

Imatinib (Gleevec) bound to ABL kinase (PDB 1IEP): a close-up that pulls the focus onto the drug,
then a tour of the pocket.

<img src="https://raw.githubusercontent.com/denphi/speck/master/media/videos/focus.gif" width="360"/>

Needs `ipyspeck>=0.8.4` and a recent Chrome, Edge, Safari (16.4+) or Firefox (130+), which make the video. Run the cells from the top, one at a time.

**Step 1: show the complex**, with the drug highlighted and a glossy look.

```python
from ipyspeck import Speck

drug = {"resName": "STI", "chain": "A"}      # imatinib in chain A; see the PDB entry for residue names

w = Speck.from_pdb_id("1IEP", cartoon=True, highlight=drug, highlightScale=1.3)
w.apply_preset("glossy")
w
```

**Step 2: the close-up.** `target=` says what to look at (a residue name, chain, residue number...).
Without it, the video visits the largest ligand.

```python
w.preview("focus", target=drug)
```

```python
w.save_video("imatinib_focus.mp4", "focus", target=drug, title="Imatinib", subtitle="bound to ABL kinase")
```

**Step 3: a longer tour** of the pocket, on a dark background with darker corners.

```python
w.save_video("imatinib_tour.mp4", "tour", target=drug, seconds=16,
             background=["#1b2330", "#07090d"], vignette=0.3)
```

### Other targets

- A whole chain: `target={"chain": "B"}`
- An active-site residue: `target={"chain": "A", "resSeq": 315}`
- Several residues: `target={"chain": "A", "resSeq": [253, 315, 381]}`

### If something does not work

- **Nothing happens after `save_video`**: the viewer must be on screen (scroll up to it) while the
  video is made; its progress bar shows there. The file appears when the bar finishes.
- **"cannot encode video"**: use a recent Chrome, Edge, Safari or Firefox, or save a GIF instead:
  `w.save_video("movie.gif", "spin")` (install what GIFs need with `pip install "ipyspeck[video]"`).
- **Too slow**: add `quality="draft"` for a quick test, or use `size="720p"`.
- **To stop a video being made**, press **Cancel** in the viewer.
